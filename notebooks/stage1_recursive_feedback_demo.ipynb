# Stage 1 recursive acoustic feedback proof-of-concept

This notebook demonstrates the minimal workflow: load a mono input WAV and a mono measured impulse response, run the fixed-gain playback path and the normalised analysis path, and save WAV/FFT/spectrogram outputs for every generation.

```python
from pathlib import Path
import sys

repo_root = Path.cwd().resolve()
if repo_root.name == 'notebooks':
    repo_root = repo_root.parent
sys.path.insert(0, str(repo_root))

from src.recursive_loop import (
    run_fixed_gain_recursive_playback,
    run_normalised_recursive_analysis,
)
```

```python
input_wav = repo_root / 'data' / 'raw_speech.wav'
impulse_response_wav = repo_root / 'data' / '35x15y.wav'
output_dir = repo_root / 'data' / 'outputs' / 'notebook_stage1'

generations = 2
max_output_seconds = 8  # keeps this example quick and compact
```

## Path A — fixed-gain recursive playback

```python
fixed_outputs = run_fixed_gain_recursive_playback(
    input_wav=input_wav,
    impulse_response_wav=impulse_response_wav,
    output_dir=output_dir,
    generations=generations,
    loop_gain=0.35,
    max_output_seconds=max_output_seconds,
)
fixed_outputs
```

## Path B — normalised recursive analysis

```python
normalised_outputs = run_normalised_recursive_analysis(
    input_wav=input_wav,
    impulse_response_wav=impulse_response_wav,
    output_dir=output_dir,
    generations=generations,
    loop_gain=1.0,
    normalisation='peak',
    normalisation_level=0.95,
    max_output_seconds=max_output_seconds,
)
normalised_outputs
```

Generated files are written under `data/outputs/notebook_stage1/`, split into `fixed_gain/` and `normalised/` subdirectories. Each path contains `wav/`, `fft/`, and `spectrogram/` folders.
